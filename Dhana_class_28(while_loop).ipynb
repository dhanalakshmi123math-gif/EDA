### When to use a while loop
- If we do not know the stop condition in advance
- If we want to give infinite chances until the answer becomes correct
- Then we can use a while loop

```python
In a for loop, the stop value is inherently available
Examples:
range(3) → stop value is 2
range(3, 11) → stop value is 

# Unknown stopping point → while loop

# Known stopping point → for loop
```
