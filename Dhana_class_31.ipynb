```python
def extract(str1):
    i1_dot=str1.index('.')
    i2_dot = str1.index('.',i1_dot+1)
    i_at = str1.index('@')
    fname=str1[:i1_dot]
    sname=str1[i1_dot+ 1: i_at]
    tname=str1[i_at+1:i2_dot]
    print(fname,sname,tname)
extract('dhana.sri@cd.com')
```
