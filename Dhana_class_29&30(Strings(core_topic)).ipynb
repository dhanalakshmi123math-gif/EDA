```python
# Q1) Ans:- 
str1 = "hai hai hai"
n=len(str1)
for i in range(n):
    # print(n-i,str1[i],i)
    if str1[i]=='a':
        print(i)

# step-1:  i=0  if str1[0]=='a'   'h'=='a'   False
# step-2:  i=1  if str1[1]=='a'   'a'=='a'   True
```
