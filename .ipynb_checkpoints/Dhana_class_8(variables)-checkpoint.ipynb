```python
- Variables are case-sensitive (e.g., number and NUMBER are different).
•	Variable names can contain both capital and small letters.
•	Numbers are not allowed as the starting character of a variable name (e.g., 123number is invalid).
•	Variable names do not include special characters (e.g., number@ is invalid).
•	Keywords or reserved words cannot be used as variable names (e.g., if, else, for, True, False, elif, def, etc.).
```
