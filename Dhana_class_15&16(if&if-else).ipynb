# Q5) WAP ask
# ask the user enter distance in km (50km)
# if distance greater than 25km
#     ask the user enter the charge per km
#     print total charge
# otherwise
#     print enjoy free ride

# idea
# step-1: distance = eval(input())
# step-2: if <condition>
# step-3:     charge = eval()
# step-4:     total charge = distance travelled * charge
# step-5: else:
# step-6:     print free ride

```python
traval_distance=eval(input('enter the distance:'))
if traval_distance>25:
    remaining_distance=traval_distance-25
    charge_per_km=eval(input('enter charge:'))
    traval_charge=remaining_distance*charge_per_km
    print(f'total price: {traval_charge}')
else:
    print(f'the enjoy free ride')
```

# Q6) WAP ask the user enter a gender

#     ask the distance
#     enter the charge
#     print the total charge

```python
gender=input('enter a gender:')
distance=eval(input('enter the distance:'))
charge=eval(input('enter the charge'))
total_charge=distance*charge
if gender=='male':
    print(f'total charge is {total_charge}')
else:
    print(f'free ride')
```
