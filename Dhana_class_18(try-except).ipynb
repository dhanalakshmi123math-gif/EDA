**try-except**



```python
There are two types
    - try block
    - except block
- write your code in try block
- if any error in try block(original code) that code will catch in except box
```

```python
a=10 
b=20 
c=a+b
print(f'sum of {a} and {b} is: {c}')
```

```python
try:
   a=10 
   b=20 
   c=a+b
   print(f'sum of {a} and {b} is: {c}') 
except Exception as e:
    print(e)
```

```python
try:
   a=10 # correct
   print(a) # correct
   b=20 # correct
   print(b)  # correct
   c=a/0 # wrong 
   print(f'sum of {a} and {b} is: {c}') 
except Exception as virat:
    print(virat)
```

```python
try:
    # wap ask the user take the salary, percentage of tax calculate the total
    bill_amount=1000
    tip_per=10
    tip_amount=bill_amount*tip_per/100
    total_amount=bill_amount+tip_amount
    print(total_amount)
except Exception as hai:
    print(hai)
```

```python
try:
    salary=20000
    tax_per=30 # per means percentage
    tax_amount=salary*tax_per/100
    total_amount=salary+tax_amount
    print(f"total_amount is {total_amount}")
except Exception as a:
    print(a)
finally: 
    print('logout')
```
