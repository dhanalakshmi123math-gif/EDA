case-1: range(stop) = range(5) 
case-2: range(start,end) = range(5,10) 
case-3: range(start,end,step) = range(1,10,2) # step means either increment or decrement

**case-3: range(start,end,step)**
- step: positive or negative value
    - range(1,10,2): step=2 positive 
    - range(1,10,-2): step=-2 negative
- start=start
- if we have positive step
    - last=stop-1
- if we have negative step
    - last=stop+1

- Example: range(1,10,2)
- start=1,  step=2  last=10-1=9
- 1   3   5  7  9

```python
for i in range(2,20,3):
    print(i,end=' ')
  # last=20-1=19 
```

```python
for i in range(2,20,-3):
    print(i,end=' ')
  # last=20+1=21 
# no answer
```

```python
for i in range(20,2,-3):
    print(i,end=' ')
  # last=2+1=3 in negative direction
```

```python
for i in range(-20,2,-3):
    print(i,end=' ')
  # last=20+1=21
```
