```python
num1=eval(input())
num2=eval(input())
print(num1,num2)
print("enter 1 for addition,\n enter 2 for subtraction,\n enter 3 for multiplication,\n enter 4 for division")
operation=eval(input('enter a number between 1 to 4')) 
if operation==1:
    print('addition operation selected')
    add=num1+num2
    print(f"sum of {num1} and {num2} is {add}")
elif operation==2:
    print('subtraction operation selected')
    sub=num1-num2
    print(f"sub of {num1} and {num2} is {sub}")
elif operation==3:
    print('multiplication operation selected')
    mul=num1*num2
    print(f"mul of {num1} and {num2} is {mul}")
elif operation==4:
    print('division operation selected')
    div=num1 / num2
    print(f"div of {num1} and {num2} is {div}")
else:
    print('enter a valid operation')
```

```python
num1=eval(input("enter a number1"))
num2=eval(input("enter a number2"))
print(num1,num2)
print("enter + for add,\n enter - for sub,\n enter * for mul,\n enter / for div")
operation=eval(input('enter a symbol [+,-,*,/]')) 
if operation=='+':
    print('addition operation selected')
    add=num1+num2
    print(f"sum of {num1} and {num2} is {add}")
elif operation=='-':
    print('subtraction operation selected')
    sub=num1-num2
    print(f"sub of {num1} and {num2} is {sub}")
elif operation=='*':
    print('multiplication operation selected')
    mul=num1*num2
    print(f"mul of {num1} and {num2} is {mul}")
elif operation=='/':
    print('division operation selected')
    div=num1 / num2
    print(f"div of {num1} and {num2} is {div}")
else:
    print('enter a valid operation')
```
