**Concept - 3: Local and Global variables**
- The variables used outside the function is called as **global variable**
- The variables used outside the function is called as **local variable**
- global variables does not have any restriction, it can go anywhere in the entire code
- local variables can not use outside the function

```python
def ADD():
    a1=100
    a2=200
    a3=300
    Addition=a1+a2+a3
    print(f"the addition of {a1},{a2} and {a3} is: {Addition}")

ADD()

# a,b,c are global and Addition is local variable
```
